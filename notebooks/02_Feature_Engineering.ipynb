```python
import pandas as pd
import numpy as np
```

```python
customers = pd.read_excel("../data/raw/Dataset1_Customers.xlsx")
transactions = pd.read_excel("../data/raw/Dataset2_Transactions.xlsx")
wallet = pd.read_excel("../data/raw/Dataset3_Wallet.xlsx")
sessions = pd.read_excel("../data/raw/Dataset4_Sessions.xlsx")
campaigns = pd.read_excel("../data/raw/Dataset5_Campaigns.xlsx")
```

```python
print("Customers:", customers.shape)
print("Transactions:", transactions.shape)
print("Wallet:", wallet.shape)
print("Sessions:", sessions.shape)
print("Campaigns:", campaigns.shape)
```

```python
print("Customers:")
print(customers.columns.tolist())

print("\nTransactions:")
print(transactions.columns.tolist())

print("\nWallet:")
print(wallet.columns.tolist())

print("\nSessions:")
print(sessions.columns.tolist())

print("\nCampaigns:")
print(campaigns.columns.tolist())
```

```python
# ==========================================
# FEATURE ENGINEERING
# IDENTIFY AVAILABLE FEATURES
# ==========================================

print("CUSTOMERS")
print(customers.columns.tolist())

print("\nTRANSACTIONS")
print(transactions.columns.tolist())

print("\nWALLET")
print(wallet.columns.tolist())

print("\nSESSIONS")
print(sessions.columns.tolist())

print("\nCAMPAIGNS")
print(campaigns.columns.tolist())
```

```python
print(
    "Customers IDs in Transactions:",
    transactions['Customer_ID'].isin(customers['Customer_ID']).sum()
)
```

```python
# ==========================================
# CUSTOMER TRANSACTION FEATURES
# ==========================================

transaction_features = transactions.groupby('Customer_ID').agg(
    total_transactions=('Transaction_ID', 'count'),
    total_spent=('Purchase_Amount', 'sum'),
    average_purchase=('Purchase_Amount', 'mean')
).reset_index()

transaction_features.head()
```

```python
transaction_features.shape
```

```python
transaction_features.info()
```

```python
transaction_features.isnull().sum()
```

```python
# ==========================================
# CUSTOMER SESSION FEATURES
# ==========================================

session_features = sessions.groupby('Customer_ID').agg(
    total_sessions=('Session_ID', 'count'),
    total_pages_visited=('Pages_Visited', 'sum')
).reset_index()

session_features.head()
```

```python
session_features.shape
```

```python
session_features.info()
```

```python
session_features.isnull().sum()
```

```python
# ==========================================
# CUSTOMER WALLET FEATURES
# ==========================================

wallet_features = wallet.groupby('Customer_ID').agg(
    total_earned_points=('Earned_Points', 'sum'),
    total_redeemed_points=('Redeemed_Points', 'sum')
).reset_index()

wallet_features.head()
```

```python
wallet_features.shape
```

```python
wallet_features.info()
```

```python
wallet_features.isnull().sum()
```

```python
campaigns.columns.tolist()
```

```python
# ==========================================
# COMBINE CUSTOMER FEATURES
# ==========================================

feature_data = customers.copy()

feature_data.head()
```

```python
feature_data = feature_data.merge(
    transaction_features,
    on='Customer_ID',
    how='left'
)
```

```python
feature_data = feature_data.merge(
    session_features,
    on='Customer_ID',
    how='left'
)
```

```python
feature_data = feature_data.merge(
    wallet_features,
    on='Customer_ID',
    how='left'
)
```

```python
feature_data.shape
```

```python
feature_data.head()
```

```python
feature_data.info()
```

```python
feature_data.isnull().sum()
```

```python
# ==========================================
# HANDLE MISSING VALUES
# ==========================================

feature_data.isnull().sum()
```

```python
feature_data.isnull().sum()[feature_data.isnull().sum() > 0]
```

```python
feature_columns = [
    'total_transactions',
    'total_spent',
    'average_purchase',
    'total_sessions',
    'total_pages_visited',
    'total_earned_points',
    'total_redeemed_points'
]

feature_data[feature_columns] = feature_data[feature_columns].fillna(0)
```

```python
feature_data[feature_columns].isnull().sum()
```

```python
feature_data.head()
```

```python
feature_data.shape
```

```python
feature_data.isnull().sum()
```

```python
# ==========================================
# SAVE PROCESSED FEATURE DATA
# ==========================================

feature_data.to_csv(
    "../data/processed/feature_data.csv",
    index=False
)
```
