```python
# Customers
for col in customers.select_dtypes(include='object').columns:
    print("\nCustomers -", col)
    print(customers[col].value_counts())

# Transactions
for col in transactions.select_dtypes(include='object').columns:
    print("\nTransactions -", col)
    print(transactions[col].value_counts())

# Wallet
for col in wallet.select_dtypes(include='object').columns:
    print("\nWallet -", col)
    print(wallet[col].value_counts())

# Sessions
for col in sessions.select_dtypes(include='object').columns:
    print("\nSessions -", col)
    print(sessions[col].value_counts())

# Campaigns
for col in campaigns.select_dtypes(include='object').columns:
    print("\nCampaigns -", col)
    print(campaigns[col].value_counts())
```
